# $M/G/1$ and Pollaczek–Khinchin

If service times are *generally* distributed we cannot use a birth-death
process. The tool is mean residual service time, and it yields the
Pollaczek–Khinchin formula: expected wait for any service-time
distribution with finite variance, $$
\bar{W}=\frac{\lambda\langle X^2\rangle}{2(1-\rho)},
$$ where $\rho=\lambda\bar{X}$ and $\bar{X}=\langle X\rangle$.

## Assumptions

The derivation uses three assumptions. The queue is FIFO. Arrival times
are statistically independent of service times. Service times have
finite variance. (The notes originally wrote \`\`arrival-time
variance’’; the 2005 paper, and the residual-time argument itself,
require finite *service-time* variance.)

Mean residual service time $\bar{R}$ is the leftover service seen by an
arrival. Under those assumptions $$
\bar{W}=\frac{\bar{R}}{1-\rho}.
$$ The rest of the argument is an expression for $\bar{R}$.

## Residual Time from Completions

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/residualTime.png" style="width:80%; height:auto; display:block;">

Figure: <i>Residual service time over a busy period: a sawtooth of
triangles with sides $X_i$. $M(t)$ is the number of completions in
$[0,t]$.</i>

The area under the residual-time curve up to $t$ is $$
A(t)=\frac12\sum_{i=1}^{M(t)}X_i^2.
$$ The time average is $$
\frac{A(t)}{t}=\frac12\frac{M(t)}{t}\frac{\sum_{i=1}^{M(t)}X_i^2}{M(t)}.
$$ As $t\to\infty$, $M(t)/t\to\lambda$ and the sample second moment
converges (finite variance), so $$
\bar{R}=\frac12\lambda\langle X^2\rangle.
$$ Substitute into $\bar{W}=\bar{R}/(1-\rho)$ to recover
Pollaczek–Khinchin. Sojourn time is then $\bar{T}=\bar{X}+\bar{W}$, and
Little gives $\bar{N}_Q=\lambda\bar{W}$ and $\bar{N}=\lambda\bar{T}$.
This derivation was Question 1 of the 2002–03 paper.

## Residual Time from an Arrival

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/residualServiceOverview.png" style="width:75%; height:auto; display:block;">

Figure: <i>The $i$th arrival sees residual $R_i$ on the customer in
service and waiting times $X$ for each of the $k_i-1$ customers already
queued.</i>

Write $W_i$ for the wait of the $i$th customer, $R_i$ for the residual
it sees, $X_i$ for its own service, and $k_i$ for the occupancy on
arrival. Then $$
W_i=R_i+\sum_{j=1}^{k_i-1}X_{(j)}.
$$ Average over the first $I$ arrivals and let $I\to\infty$. Time
averages equal steady-state means, and independence of the $i$th arrival
from the work already present, together with Little on the queue,
recover $\bar{W}=\bar{R}/(1-\rho)$.

# Special Cases

Second moment $\langle X^2\rangle=\mathrm{Var}(X)+\bar{X}^2$. For
exponential service, $\bar{X}=1/\mu$ and $\mathrm{Var}(X)=1/\mu^2$, so
$\langle X^2\rangle=2/\mu^2$ and Pollaczek–Khinchin collapses to the
$M/M/1$ wait $\bar{W}=\rho/(\mu(1-\rho))$.

Deterministic service ($M/D/1$, fixed-length packets on a
fixed-bandwidth line) has $\mathrm{Var}(X)=0$, so
$\langle X^2\rangle=1/\mu^2$ and $$
\bar{W}=\frac{\rho}{2\mu(1-\rho)},
$$ half the $M/M/1$ wait. Among distributions with a given mean,
deterministic service gives the lowest delay: the residual sawtooth has
the smallest area.

Uniform service on $[0,b]$ has mean $b/2$ and variance $b^2/12$. With
$\mu=2/b$, the second moment is $4/(3\mu^2)$ and $$
\bar{W}=\frac{2\rho}{3\mu(1-\rho)}.
$$

Writing $\mu=1/\bar{X}$: exponential wait $\rho/(\mu(1-\rho))$, sojourn
$1/(\mu(1-\rho))$, $\bar{N}=\rho/(1-\rho)$; deterministic wait half of
that, sojourn $(2-\rho)/(2\mu(1-\rho))$,
$\bar{N}=\rho(2-\rho)/(2(1-\rho))$.

# $M/G/1$ with Vacations

When the server is free it may attend to other work — control traffic, a
kitchen check — a *vacation*. An arrival that finds the server on
vacation waits until the vacation ends. If a vacation ends and the
server is still idle, it takes another. Vacation lengths
$V_1,V_2,\ldots$ are i.i.d. and independent of arrivals and service.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/vacationMG1.png" style="width:70%; height:auto; display:block;">

Figure: <i>An $M/G/1$ server that takes a vacation whenever the queue is
empty.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/vacationResidual.png" style="width:70%; height:auto; display:block;">

Figure: <i>Residual time now includes vacation triangles as well as
service triangles.</i>

The residual-time integral now has two families of triangles. With
$M(t)$ service completions and $L(t)$ vacations in $[0,t]$, $$
\frac1t\int_0^t r(\tau)\,d\tau
=\frac{M(t)}{t}\frac12\frac{\sum X_i^2}{M(t)}
+\frac{L(t)}{t}\frac12\frac{\sum V_i^2}{L(t)}.
$$ The fraction of time available for vacations is $1-\rho$, so
$L(t)=t(1-\rho)/\bar{V}_t$. The limit is $$
\bar{R}=\frac12\lambda\langle X^2\rangle+\frac{(1-\rho)\langle V^2\rangle}{2\bar{V}},
$$ and therefore $$
\bar{W}=\frac{\lambda\langle X^2\rangle}{2(1-\rho)}+\frac{\langle V^2\rangle}{2\bar{V}}.
$$ Pollaczek–Khinchin plus a vacation residual that does *not* depend on
$\rho$.

A bar with Poisson arrivals, exponential service of mean $\bar{X}$, and
exponential kitchen checks of mean $\bar{V}$ has
$\langle X^2\rangle=2\bar{X}^2$ and $\langle V^2\rangle=2\bar{V}^2$, so
$$
\bar{W}=\frac{\rho\bar{X}}{1-\rho}+\bar{V}.
$$ Deterministic vacations of the same mean would add only $\bar{V}/2$.
