# Erlang Delay: $M/M/m$

$M/M/\infty$ is the unlimited-server extreme. A call centre, a trunk
group with holding, or any pool of $m$ identical exponential servers
with a waiting line is the *Erlang delay* system: Kendall $M/M/m$.
Arrivals that find all servers busy join an unbounded queue. The system
is lossless, so carried load equals offered load $a=\lambda/\mu$. The
mean holding time is $\tau=1/\mu\equiv\bar{X}$.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/erlangDelayDiagram.png" style="width:70%; height:auto; display:block;">

Figure: <i>Schematic of Erlang delay: offered load $a$ enters through a
queue; nothing is lost, so carried load equals $a$.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/erlangDelay.png" style="width:85%; height:auto; display:block;">

Figure: <i>State diagram of $M/M/m$: death rate climbs as $k\mu$ until
$k=m$, then saturates at $m\mu$.</i>

Birth-death rates are $$
\lambda_k=\lambda,\qquad
\mu_k=\begin{cases}
k\mu & 1\le k\le m,\\
m\mu & k>m.
\end{cases}
$$ Utilisation of the *server pool* is $\rho=\lambda/(m\mu)=a/m$.
Stability still requires $\rho<1$, i.e. $a<m$: the arrival rate must be
less than the full-pool service rate.

## Detailed Balance

For $k\le m$ the boundary is the $M/M/\infty$ boundary, $$
\lambda P(k-1)=k\mu P(k)\qquad\Rightarrow\qquad P(k)=\frac{m\rho}{k}P(k-1)=\frac{(m\rho)^k}{k!}P(0).
$$

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/boundaryFlowMMinf.png" style="width:45%; height:auto; display:block;">

Figure: <i>Boundary flow while at least one server is free:
$\lambda P(k-1)=k\mu P(k)$.</i>

For $k>m$ every server is busy, so the death rate is $m\mu$, $$
\lambda P(k-1)=m\mu P(k)\qquad\Rightarrow\qquad P(k)=\rho P(k-1)=\rho^{k-m}P(m).
$$ With $P(m)=(m\rho)^m P(0)/m!$ this is $$
P(k)=\frac{m^m\rho^k}{m!}P(0),\qquad k>m.
$$

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/boundaryFlowMMm.png" style="width:45%; height:auto; display:block;">

Figure: <i>Boundary flow once all $m$ servers are busy:
$\lambda P(k-1)=m\mu P(k)$.</i>

The stationary distribution is therefore piecewise $$
P(k)=\begin{cases}
P(0)\,(m\rho)^k/k! & k\le m,\\
P(0)\,m^m\rho^k/m! & k>m.
\end{cases}
$$ Normalisation, summing the Poisson-like head and the geometric tail,
gives $$
P(0)=\left[\sum_{k=0}^{m-1}\frac{(m\rho)^k}{k!}+\frac{(m\rho)^m}{m!(1-\rho)}\right]^{-1}.
$$

# Probability of Waiting

The operational question for $M/M/m$ is the probability that an arrival
finds every server busy and must wait, $$
P_Q=P(k\ge m)=\sum_{k=m}^{\infty}P(k)=P(0)\frac{(m\rho)^m}{m!(1-\rho)}\doteq C(m,a).
$$ This is *Erlang’s second formula*, or Erlang C. In offered-load form
$a=m\rho$, $$
C(m,a)=\frac{P(0)\,a^m}{m!(1-a/m)}.
$$

Direct evaluation of $a^m/m!$ overflows for large $m$. Compute in log
space, $$
\ln C(m,a)=\ln P(0)+m\ln a-\ln m!-\ln(1-a/m),
$$ with $\ln m!=\ln\Gamma(m+1)$. The original MATLAB helpers `erlangC.m`
/ `plotErlang.m` do that; `npa_plots.erlang_c` and
`npa_plots.plot_erlang_curves` are the same calculation.

# Mean Occupancy and Waiting Time

The mean number in the system splits into the mean queue length and the
mean number of busy servers $a=m\rho$, $$
\bar{N}=\frac{\rho}{1-\rho}C(m,a)+m\rho,\qquad
\bar{N}_Q=\frac{\rho}{1-\rho}C(m,a).
$$ Little’s formula on the waiting line, with arrival rate $\lambda$
(nothing is lost), gives the mean wait for *all* customers, $$
\bar{W}=\frac{\bar{N}_Q}{\lambda}=\frac{C(m,a)}{m\mu(1-\rho)}.
$$ Customers who do not queue have $\bar{W}_{\sim Q}=0$, so the mean
wait among those who *do* queue is $$
\bar{W}=C(m,a)\,\bar{W}_Q\qquad\Rightarrow\qquad
\bar{W}_Q=\frac{1}{m\mu(1-\rho)}=\frac{1}{\mu(m-a)}.
$$ That conditional wait does not depend on $C(m,a)$.
