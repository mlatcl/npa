# Introduction

Welcome to Network Performance Analysis: sixteen lectures and two lab
sessions on queueing theory and network simulation.

Queueing theory dates from the beginning of the twentieth century and
the work of **Agner Krarup Erlang** (1878–1929), a Danish telephone
engineer who published the first paper on the subject (Erlang, 1909).

The first two chapters place that theory in the development of worldwide
communication networks. For further reading see
[Wikipedia](http://en.wikipedia.org) and the [Computer History
Museum](http://www.computerhistory.org).

# Early Milestones

The first major development in telecommunications was the *telegraph*.
**Samuel F. B. Morse** built an electrical telegraph that exploited the
link between magnetism and electricity. His first working system sent a
short message from Baltimore to Washington, D.C. on 24 May 1844: “What
hath God wrought?” Ten years later there were 23,000 miles of wire
across the United States.

A key limitation of the telegraph was that it provided only one channel
per wire. Work on using a single wire for multiple channels
(multiplexing) led indirectly to the *telephone*, patented by
**Alexander Graham Bell** (1847–1922) on 7 March 1876.

**Guglielmo Marconi** (1874–1937) developed wireless telegraphy,
patented in July 1897, and founded the Wireless Telegraph and Signal
Company. By December 1901 short-wave transmission crossed the Atlantic
between England and Canada (over 2,100 miles).

# Automated Telephone Networks

Early telephone routing was manual. Operators at the exchange used patch
boards to connect callers. **Almon Strowger** (1839–1902) invented the
automatic telephone exchange (the Strowger switch) in 1891, and that
created a need for a theoretical study of how exchanges behave under
demand.

The network of interconnected phone lines is a *circuit switching*
network: lines are switched between two subscribers to form a
connection. Erlang’s contribution was to give theoretical studies of
these networks that allowed exchanges to be designed to meet demand.

# Information Theory

As telephone networks grew, so did the need for a theory of what they
carry. Networks transmit “information”, but talk of channel capacity is
empty without a definition. **Claude E. Shannon** (1916–2001) wrote *A
Mathematical Theory of Communication* (Shannon, 1948), which founded
information theory and gave a mathematical meaning to the capacity of a
channel. We will not cover that theory here.

# Computer Networks

Circuit-switched networks were designed for a telephone conversation
between two people. They were the first worldwide communication
networks. By the early 1960s a different kind of network was needed. In
circuit switching a permanently open line connects the correspondents
and a fixed bandwidth is reserved for them. If they do not use it,
nobody else can.

In *packet switching* (store-and-forward) the bandwidth is shared. What
a user sends — voice over IP, or a file by FTP — is split into packets
that are routed across the network. If a required line is busy the
packet waits in a queue. The ideas were developed independently by
**Paul Baran** (1926–2011) at RAND and **Donald Davies** (1924–2000) at
the National Physical Laboratory. **Leonard Kleinrock**’s 1962 thesis,
later a book (Kleinrock, 1964), gave the mathematical theory that
underpins these networks.

## ARPANET

ARPA funded a network that implemented these ideas. In 1969 it connected
four nodes: UCLA, the Stanford Research Institute, UC Santa Barbara, and
the University of Utah. By 1975 the network covered most major US
universities (61 nodes) and a node at University College London.

Figure: <i>The ARPANET in 1975, image courtesy of the Computer History
Museum. PDP machines are DEC minicomputers; 360 and 370 are IBM

ARPANET was restricted to non-commercial use and funded by ARPA and NSF
until 1990. Telnet and FTP appeared in 1971. The transmission control
protocol (TCP), developed in the mid-1970s by **Vinton Cerf** (1943–)
and **Robert Kahn** (1938–) (Cerf and Kahn, 1974), allowed
interconnection between networks and thereby the internet.

## The Internet

By 1977 Cerf and Kahn could send messages across the Packet Radio net,
SATNET, and ARPANET. At a similar time **Robert Metcalfe** (1946–) was
developing Ethernet at Xerox PARC. Ethernet was inspired by the ALOHA
networks at the University of Hawaii: machines shared a radio channel,
collisions had to be detected, and packets retransmitted. Ethernet used
a cable rather than a radio frequency. Cheap Ethernet and cheap
computers produced an explosion in the number of networks on the
internet.

When ARPANET was shut down in 1990 it had grown to 300,000 hosts. It
evolved into NSFNet, the backbone of the internet. In 1991 the internet
opened to commercial traffic. Combined with HTML and URLs, invented by
**Tim Berners-Lee** (1955–), that is the foundation of the modern
internet.

# Network Performance Analysis

This course introduces the tools for design and analysis of
telecommunication networks. The core tools are probabilistic: we assume
that traffic is stochastic.

The same techniques apply beyond telecommunications. Queues form at
compute servers, in databases handling queries, at cashpoints, in
supermarkets, and in telephone call centres. The principles apply in
each of those settings.

## Design and Quality

For network design we care about two related questions.

**Design issues.** When building or buying a router, how much memory
should we allow? How much bandwidth do we need between routers? How much
processing power should the router have so that it can use that
bandwidth?

**Quality issues.** What will be the average delay for end users given
our design? What is the worst-case delay? How likely is the system to
drop packets?

## Terminology

A series of communications between two users is a *session*. The
interactions consist of *messages*, each of which may be split into
*packets*.

One example of a session is a lecture: mainly one-way communication from
lecturer (server) to students (clients). Messages cover different
subjects; packets might be the sentences. The medium is air.

## Session Characteristics

FTP, VoIP, and HTTP sessions look very different. Six characteristics of
a session matter for protocol and network design.

This course concentrates on the first five.

## Circuit Switching and Packet Switching

We have already met two kinds of network.

**Circuit switching** can waste bandwidth, because the two parties may
not use the reserved channel all the time. Once connected, though, delay
is known. The sum of session rates on a link cannot exceed the link
capacity; new sessions are rejected when capacity is reached. That is
natural for telephony, where each session has the same rate $r_s$.

**Packet switching** uses bandwidth more efficiently: idle time on a
conversation can carry someone else’s packets. Delay is then harder to
guarantee.

Figure: <i>A simple circuit-switching network connecting Alice and Bob.
The session dominates both phone lines.</i>

Alice calls Bob on the phone: the session occupies both lines.
Alternatively they use an instant-messenger voice session while Bob’s
friend Carol shares the land line to reach David’s web server.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/aliceBobPacket.png" style="width:70%; height:auto; display:block;">

Figure: <i>A simple packet-switching network connecting Alice, Bob,
Carol, and David. Congestion on the shared land line produces variable

A router in Bob’s house now shares the land line. Bob and Carol can both
use it; if Carol downloads a large file, Bob sees a variable delay. That
queueing delay is the object of this course.

## Communication Media

Circuit-switched networks share a line by multiplexing: time-division or
frequency-division. Neither is perfectly efficient. Packet-switched
networks give bandwidth to a session as required. In a simple FIFO
system a packet is transmitted if the queue is empty; otherwise it
waits. There are two delays: a variable queueing delay, and a
transmission delay equal to packet length times the reciprocal of the
line rate.

Radio goes back to Marconi. The most basic wireless packet networks
(ALOHA) used one channel. If two stations transmit at once, both packets
are lost. Thinwire Ethernet has the same collision structure, with a
coaxial cable as the medium.

## Example: Matching Disk and CPU

A simple design question. We are building database servers from existing
parts: two CPU/motherboard systems and two disk systems.

| CPU system | Expected service time (ms) | Disk system | Expected service time (ms) |
|:----------|:-----------------------|------------|:-----------------------|
| A          | 423                        | 1           | 215                        |
| B          | 210                        | 2           | 390                        |

Which should we match with which? One fast system and one slow system,
or two similar systems?

Match A with 2 and B with 1. Pairing the slow CPU with the slow disk and
the fast CPU with the fast disk avoids a one-sided wait: disk 1 would
otherwise wait for CPU A on about half the jobs, and CPU B would wait
for disk 2 on about half the jobs.
