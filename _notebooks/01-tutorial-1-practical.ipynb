# Falling leaves

Leaves fall from a tree as a Poisson process at a rate of one every
minute.

1.  You arrive at 16:00. During the next ten minutes, what is the
    probability that you see twelve leaves fall?

2.  What is the probability that you see $n$ leaves fall for
    $n=0,\ldots,20$? Plot the results.

```
from scipy.stats import poisson
lam, t = 1.0, 10.0
n = np.arange(0, 21)
p = poisson.pmf(n, lam * t)
print("P(k=12) =", poisson.pmf(12, lam * t))
plt.stem(n, p)
plt.xlabel("$n$")
plt.ylabel(r"$P(n\mid\lambda t=10)$")
plt.show()
```

1.  Where does the curve peak? Compare the peak with the expected value
    $\lambda t$.

\langle n^2\rangle=\lambda t\, e^{-\lambda t}\sum_{n=1}^{\infty}\frac{n(\lambda t)^{n-1}}{(n-1)!},

# Gaussian samples

This question needs a computer. Draw $N$ samples $\{x_n\}_{n=1}^N$ from
a Gaussian with mean 0 and variance 1 (`numpy.random.randn` is the old
MATLAB `randn`). For each $N$ in the table, compute the sample mean and
the sample second moment.

| $N$                       | 10  | 50  | 100 | 200 | 500 | 1000 |
|:--------------------------|:----|:----|:----|:----|:----|:-----|
| $\frac{1}{N}\sum_n x_n$   |     |     |     |     |     |      |
| $\frac{1}{N}\sum_n x_n^2$ |     |     |     |     |     |      |

```
Ns = [10, 50, 100, 200, 500, 1000]
rng = np.random.default_rng(1)
for N in Ns:
    x = rng.normal(size=N)
    print(N, x.mean(), np.mean(x**2))
```

1.  What is the true mean of the distribution you have sampled from?

2.  For each $N$, plot the squared difference between the sample mean
    and the true mean. What sort of trend do the points give? Is that
    trend consistent? When, typically, are the sample estimates of the
    mean worst?

3.  What is the true second moment of the distribution?

4.  For each $N$, plot the squared difference between the sample second
    moment and the true second moment. Is the trend similar to that of
    the sample mean? Are the second-moment estimates typically better or
    worse? Why?

*Hint.* Repeat the experiment several times and summarise with means and
standard deviations. `matplotlib.pyplot.errorbar` replaces MATLAB
`errorbar`.

# The exponential distribution

Compute the mean and variance of $$
p(x)=\lambda e^{-\lambda x},\qquad 0<x<\infty.
$$

Repeat the sampling experiments of the previous question with an
exponential of rate $\lambda=3$, using `npa_plots.exponential_sample` in
place of the old `exponentialSamp.m`.

```
lam = 3.0
Ns = [10, 50, 100, 200, 500, 1000]
for N in Ns:
    x = npa_plots.exponential_sample(lam, size=N, rng=1)
    print(N, x.mean(), np.mean(x**2))
```
