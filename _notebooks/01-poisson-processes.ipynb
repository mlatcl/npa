# Stochastic Processes

A probabilistic process — a random process, or a stochastic process — is
a random function: it maps inputs to a distribution over values rather
than to a single value. The Wiener process sits behind Brownian motion.
The Black–Scholes formula models the logarithm of a stock price as a
Wiener process. Following (Chan, 2000), a random process is a random
function which, for a particular input, is a random quantity.

# The Poisson Process

The Poisson *distribution* gives probabilities over counts. The Poisson
*process* gives probabilities of $k$ arrivals as a function of time $t$
for a rate $\lambda$, $$
P(k\mid t,\lambda)=\frac{(\lambda t)^k}{k!}e^{-\lambda t}.
$$ This is the Poisson distribution with $\mu$ replaced by $\lambda t$.
The mean number of arrivals by time $t$ is therefore
$\langle k\rangle=\lambda t$, so $\lambda$ is a rate. The Poisson
process is a counting process whose counts are expected to grow linearly
in time.

# Properties of the Poisson Process

In the next chapter we will derive the Poisson process from assumptions
about arrivals. For now, five properties.

## Superposition

If $N$ independent Poisson processes with rates
$\lambda_1,\dots,\lambda_N$ are combined, the result is Poisson with
rate $\lambda=\sum_n\lambda_n$. Two variables are independent when
$P(x,y)=P(x)P(y)$.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/sumPoisson.png" style="width:55%; height:auto; display:block;">

Figure: <i>Independent Poisson processes merged into one flow yield a
Poisson process whose rate is the sum of the input rates.</i>

## Decomposition

The converse also holds. A Poisson process of rate $\lambda$ is split
onto $N$ output channels, each arrival independently assigned to channel
$n$ with probability $P(n)$. Then channel $n$ is Poisson with rate
$\lambda_n=P(n)\lambda$.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/decomposePoisson.png" style="width:70%; height:auto; display:block;">

Figure: <i>A Poisson process of rate $\lambda$ independently switched
onto output $n$ with probability $P(n)=\lambda_n/\lambda$ yields Poisson
outputs of rates $\lambda_n$.</i>

## Independence of Arrivals

On two *disjoint* time intervals the numbers of arrivals $k_1$ and $k_2$
are independent: $P(k_1,k_2)=P(k_1)P(k_2)$.

## Identical Intervals

For a single rate $\lambda$, the number of arrivals in $(s,s+t]$ has the
same distribution as the number in $(r,r+t]$ on an independent copy of
the process.

# Interarrival Times

The distribution of times between arrivals of a Poisson process is
exponential, $$
p(X)=\lambda e^{-\lambda X}.
$$ Stating that interarrival times are exponential with parameter
$\lambda$ is the same as stating that arrivals are Poisson with rate
$\lambda$. If students register at a rate $\lambda=2$ per minute, the
gaps between them are exponential with the same $\lambda$.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/interArrivals.png" style="width:60%; height:auto; display:block;">

Figure: <i>Arrival epochs $T_n$ and interarrival times $X_n=T_n-T_{n-1}$
for a Poisson process.</i>

The derivation uses the CDF. $P(X_i>t)$ is the probability of no
arrivals in an interval of length $t$, which is $e^{-\lambda t}$.
Differentiating $1-e^{-\lambda t}$ recovers $\lambda e^{-\lambda X}$.
That argument is written out as an example at the end of the original
notes and as an exercise below.

# The Exponential Distribution

## Memoryless

The exponential is *memoryless*: $$
P(X>s+t\mid X>t)=P(X>s).
$$ Having already waited $t$ does not change the distribution of the
remaining wait. That is the property we will use for Markov queues.

## Mean and Variance

For $p(X)=\lambda e^{-\lambda X}$, $$
\langle X\rangle=\frac{1}{\lambda}, \qquad \mathrm{Var}(X)=\frac{1}{\lambda^2}.
$$ Confirming these is part of Tutorial 1.

The MATLAB helper `exponentialSamp.m` drew $U\sim\mathrm{Uniform}(0,1)$
and returned $-\log U/\lambda$. That is the inverse-CDF sampler.

```
lam = 2.0
x = npa_plots.exponential_sample(lam, size=20_000, rng=1)
print("sample mean", x.mean(), "should be", 1/lam)
print("sample variance", x.var(), "should be", 1/lam**2)
plt.hist(x, bins=40, density=True, alpha=0.6, label="samples")
t = np.linspace(0, x.max(), 200)
plt.plot(t, lam * np.exp(-lam * t), label="exponential pdf")
plt.xlabel("interarrival time")
plt.legend()
plt.show()
```

# Worked Examples

## Superposition

Three independent Poisson registration queues: Physics and Fine Art at
rate 4 per minute, Psychology of Sociology at rate 2, Computer Science
and Aromatherapy at rate 5. Combined rate $\lambda=11$. After 15 minutes
the expected number of arrivals is $165$.

Evaluate $P(k\mid\lambda=11,t=15)$ in the log domain, $$
\log P = k\log(\lambda t)-\log(k!)-(\lambda t),
$$ because $180!$ will overflow a naive factorial.

```
from scipy.stats import poisson
lam, t = 11.0, 15.0
mu = lam * t
print("P(k=90) =", poisson.pmf(90, mu))
print("P(k=180) =", poisson.pmf(180, mu))
```

## Decomposition

One hall, four desks by first letter of surname, with
$P(d)=(0.20,0.27,0.25,0.28)$ and total rate 10 students per minute. Desk
1 is Poisson with rate $2$; $P(k=5\mid t=5)\approx 0.038$. Desk 4 has
rate $2.8$; $P(k=20\mid t=10)\approx 0.025$.

```
print("desk 1, k=5, t=5:", poisson.pmf(5, 2.0 * 5))
print("desk 4, k=20, t=10:", poisson.pmf(20, 2.8 * 10))
```
