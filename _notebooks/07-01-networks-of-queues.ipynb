# Queues in Tandem

So far every model has been a single queue. Two transmission lines in
series are the smallest network. Packet length does not change from hop
to hop, and service time is proportional to length. Arrivals at the
first queue are Poisson; lengths are exponential.

The first hop is $M/M/1$. The second is not. A long packet occupies the
first server for a long time, during which the second queue can drain. A
long packet therefore tends to arrive at the second hop when that queue
is shorter, and waits *less* there than a short packet would. The
analogy is a slow vehicle on a road: from that driver’s seat the road
looks empty, because they never catch anyone. The first queue has
destroyed independence between a packet’s length and the state of the
second queue.

That correlation is hard to treat exactly. The way through is to *break*
it: pretend service times at the two hops are statistically independent,
even though they are the same length. Lab 2 asked you to look at whether
the simulated occupancies still match isolated $M/M/1$ formulae.
Kleinrock’s independence assumption, below, is that pretence made
systematic.

# Time Reversibility and Burke

Birth-death chains (every Markov queue in the course except $M/G/1$)
satisfy detailed balance. They are also *time-reversible* in steady
state: the arrival process is Poisson at rate $\lambda$, and so is the
*departure* process. The service rate $\mu$ is not the departure rate,
because a departure can occur only when $k>0$. That happens a fraction
$\rho$ of the time, so the departure rate is $\mu\rho=\lambda$.

Burke’s theorem: an $M/M/1$, $M/M/m$, or $M/M/\infty$ queue in steady
state with Poisson arrivals at rate $\lambda$ has a Poisson departure
process at rate $\lambda$, *and* the occupancy at time $t$ is
independent of the departure times before $t$. Loss systems are
excluded: blocked customers never enter, so the process that does enter
is not rate $\lambda$.

The second claim is the surprising one. A burst of departures might seem
to leave the queue short. Burke says it does not. The only way both can
be true is that the occupancy *before* the burst was abnormally high.

If we grant independent service at the second hop, Burke says the input
to that hop is Poisson at $\lambda$. Two $M/M/1$ queues in series then
have occupancies $$
P_1(k)=(1-\rho_1)\rho_1^k,\qquad P_2(\ell)=(1-\rho_2)\rho_2^\ell,
$$ and the joint distribution factors. Mean delay is the sum of the two
sojourn times.

# Kleinrock’s Independence Assumption

Kleinrock’s independence assumption treats every link as if it saw
Poisson arrivals and independent service — as if merging streams
restored the independence that tandem service destroys. It is a good
approximation when many packets from outside the network enter at every
queue, on densely connected networks, with exponential packet lengths,
Poisson exogenous arrivals, and moderate to heavy traffic.

Label nodes and links. Stream $s$ has rate $x_s$; $f_{ij}(s)$ is the
fraction of that stream that uses the link from $i$ to $j$. The link
rate is $$
\lambda_{ij}=\sum_s f_{ij}(s)\,x_s.
$$ Each link is then an $M/M/1$ queue with mean occupancy and delay $$
\bar{N}_{ij}=\frac{\lambda_{ij}}{\mu_{ij}-\lambda_{ij}},\qquad
\bar{T}_{ij}=\frac{1}{\mu_{ij}-\lambda_{ij}}.
$$ The mean number in the network is the sum over links. We ignore
propagation and processing delay; see Bertsekas and Gallager, Section
3.6.1.

# Jackson’s Theorem

Kleinrock approximates each *link*. Jackson’s theorem is a statement
about the *network*. Assumptions: no correlation between service times
and queue lengths (Kleinrock’s premise), and routing that randomises
among paths. Then the mean number of packets in the network is $$
\bar{N}=\sum_{i,j}\frac{\lambda_{ij}}{\mu_{ij}-\lambda_{ij}}.
$$ The product-form occupancy is the deeper result; for this course the
mean and the two assumptions are enough. Analysis is in Bertsekas and
Gallager, Section 3.8. This is Question 4 of the 2004–05 paper.

$1\to 2\to 5$.

# Simulation

Kleinrock and Jackson are back-of-envelope tools. A real network is
usually simulated. The 2006 lab used ns-2, a discrete-event network
simulator driven by Tcl, supporting TCP and multicast, extensible in
C++, and installed on the departmental Linux machines (Cygwin for a
personal copy). Lab 2 is that comparison: occupancy from ns-2 against
isolated $M/M/1$ formulae on a tandem pair and a multipath diamond.

The approximations are useful *before* the simulator, and as a check on
it. Lab 1 already showed that a sample mean from a short run at high
$\rho$ is not the steady-state mean. The same warning applies to packet
traces.
