# Markov Chains

The Poisson process is a special case of a Markov process. We start in
discrete time, with *Markov chains*, then take a continuum limit. Both
chains and processes obey the Markov property: the future is
conditionally independent of the past given the present.

## The Markov Property

Let time be discrete, $t=0,1,2,\ldots$, and let $N(t)$ be a state. The
chain has the Markov property if $N(t)$ depends on the history only
through $N(t-1)$.

## Monopoly

A single-player Monopoly board has forty squares. Let $N(t)$ be the
square index plus $40$ for each completed circuit: twice round and on GO
gives $N(t)=80$; once round and on King’s Cross gives $N(t)=45$. Time
$t$ counts turns.

Each turn the player rolls two six-sided dice. Write $d(t)$ for the
total. Then $$
N(t)=N(t-1)+d(t).
$$ The next square depends only on the current square and the current
roll, so the walk has the Markov property.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/smallMonopoly.png" style="width:45%; height:auto; display:block;">

Figure: <i>A Monopoly board. The token’s square after a turn depends
only on the square before the turn and the dice.</i>

## State Diagrams

A Markov chain has a state diagram in the same spirit as a finite-state
machine, with two differences. The state space need not be finite.
Transitions carry probabilities (or, for a process, rates) rather than
deterministic labels.

# The Poisson Process as a Markov Process

We now pass from a discrete-time chain to a continuous-time process, and
recover the Poisson process as a Markov process with a restricted set of
transitions. The device is a one-dimensional walk we call Coin Monopoly.

## Coin Monopoly

Anne has lost the dice. She tosses a coin instead: $d(t)=0$ on tails and
$d(t)=1$ on heads. The state either stays put or advances by one square.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/stateDiagram.png" style="width:70%; height:auto; display:block;">

Figure: <i>State diagram for Coin Monopoly. Each node is a square index.
With each turn the token stays or moves one step, with the probabilities
on the arrows.</i>

The game is slow. Speed it up by shrinking the turn length $\Delta t$
and using a coin whose heads probability is $P(h)=\lambda\Delta t<1$, so
$P(\text{tails})=1-\lambda\Delta t$.

## Binomial Trials to a Poisson Process

After $n$ turns the number of heads $k$ is binomial, $$
P(k)=\binom{n}{k}P(h)^k\bigl(1-P(h)\bigr)^{n-k},\qquad \binom{n}{k}=\frac{n!}{k!(n-k)!}.
$$ In time $t$ there are $n=t/\Delta t$ turns, so $P(h)=\lambda t/n$.
Substitute and take $n\to\infty$, which is $\Delta t\to 0$: $$
\lim_{n\to\infty}\frac{n(n-1)\cdots(n-k+1)}{n^k\bigl(1-\lambda t/n\bigr)^k}=1,
\qquad
\lim_{n\to\infty}\Bigl(1-\frac{\lambda t}{n}\Bigr)^{n}=e^{-\lambda t}.
$$ The occupancy becomes $$
P(k\mid\lambda,t)=\frac{(\lambda t)^k}{k!}e^{-\lambda t}.
$$ That is a Poisson process: a Markov *process*, because time is now
continuous.

## Process State Diagrams

On a continuous-time diagram we drop the self-transition arrows and
replace transition probabilities by rates.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/stateDiagramProcess.png" style="width:70%; height:auto; display:block;">

Figure: <i>Coin Monopoly in the limit $\Delta t\to 0$: a Poisson
process. Rates replace probabilities, and there are no self-loops.</i>

The construction explains two Poisson properties from the last lecture.
At every instant the probability of an arrival is constant, and it does
not depend on what has already happened. That is the Markov property in
continuous time.

# Steady State

Once the Poisson process sits inside the Markov family, the next objects
are birth-death processes and the $M/M/1$ queue. We first need *steady
state*: a condition that does not change with time, or in which change
in one direction is continually balanced by change in the other. Network
delay, once the transient has died, is a steady-state question.

# Birth-Death Processes

Start from the Coin Monopoly chain and allow a step *backwards* as well
as forwards: $k\to k+1$ (a birth) and $k\to k-1$ (a death), with the
state forbidden from going negative. That is a birth-death process.
Discrete and continuous versions have the state diagrams below.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/birthDeathChain.png" style="width:70%; height:auto; display:block;">

Figure: <i>State diagram for a discrete-time birth-death chain.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/birthDeathProcess.png" style="width:70%; height:auto; display:block;">

Figure: <i>State diagram for a continuous-time birth-death process.
Births increase $k$; deaths decrease it.</i>

# Kendall’s Notation

Kendall’s notation summarises a queue as $$
A/S/m/K,
$$ arrival process / service process / number of servers / capacity
(optional). The letters for $A$ and $S$ are $M$ (Markov: Poisson
arrivals or exponential service), $D$ (deterministic), and $G$
(general).

# Global Balance

A stationary distribution satisfies *global balance*: at equilibrium the
frequency of transitions out of $k$ equals the frequency of transitions
into $k$ (Bertsekas and Gallager, 1992, pp. ~260–262).

## Discrete Time

A chain is specified by a transition matrix
$P_{ij}=\Pr(\text{next}=j\mid\text{now}=i)$. If $P(j)$ is the
steady-state probability of state $j$, global balance is $$
P(j)\sum_{i}P_{ji}=\sum_{i}P(i)P_{ij}:
$$ the probability of leaving $j$ equals the probability of arriving in
$j$.

## Continuous Time

A process is specified by rates $q_{ij}$ from $i$ to $j$. Global balance
becomes $$
P(j)\sum_{i}q_{ji}=\sum_{i}P(i)q_{ij}.
$$

# Detailed Balance

In a birth-death process the only transitions are between neighbours, so
$P_{ij}=0$ unless $|i-j|\le 1$. Global balance then reduces to *detailed
balance* between $k-1$ and $k$: $$
P(k-1)\,P_{k-1,k}=P(k)\,P_{k,k-1},
$$ or, with rates $\lambda_{k-1}$ (birth from $k-1$) and $\mu_k$ (death
from $k$), $$
P(k-1)\lambda_{k-1}=P(k)\mu_k.
$$

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/boundaryFlow.png" style="width:55%; height:auto; display:block;">

Figure: <i>Flow across the boundary between $k-1$ and $k$. At
equilibrium $\lambda_{k-1}P(k-1)=\mu_k P(k)$.</i>

Rearrangement gives the recurrence $$
P(k)=\frac{\lambda_{k-1}}{\mu_k}P(k-1).
$$ Different queues are different choices of $\lambda_k$ and $\mu_k$.
The $M/M/1$ queue takes both rates constant.

# The $M/M/1$ Queue

Kendall’s $M/M/1$ is Poisson arrivals at rate $\lambda$ and exponential
service at rate $\mu$: a birth-death process with $\lambda_k=\lambda$
and $\mu_k=\mu$ for every $k\ge 1$, and no death from $0$.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/stateMM1.png" style="width:70%; height:auto; display:block;">

Figure: <i>State diagram for the $M/M/1$ queue: unbounded $k$, no
transition below $0$, constant rates.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/boundaryFlowMM1.png" style="width:55%; height:auto; display:block;">

Figure: <i>Boundary flow for $M/M/1$: $\lambda P(k-1)=\mu P(k)$.</i>

Detailed balance gives $$
P(k)=\frac{\lambda}{\mu}P(k-1)=\rho^k P(0),\qquad \rho=\frac{\lambda}{\mu}.
$$ The ratio $\rho$ is the *utilisation*. Normalisation over
$k=0,1,2,\ldots$ requires $$
\sum_{k=0}^{\infty}P(k)=1=P(0)\sum_{k=0}^{\infty}\rho^k.
$$ The geometric series converges only for $\rho<1$, $$
\sum_{k=0}^{\infty}\rho^k=\frac{1}{1-\rho},
$$ so $P(0)=1-\rho$ and $$
P(k)=\rho^k(1-\rho).
$$ If $\rho\ge 1$ there is no steady state: the queue grows without
bound.

## Mean Occupancy

The mean number in the *system* (queue plus server) is $$
\langle k\rangle=\sum_{k=0}^{\infty}kP(k)=\frac{\rho}{1-\rho}=\frac{\lambda}{\mu-\lambda}.
$$ That is not the mean queue length: a customer in service is in the
system but not waiting. The occupancy variance is $\rho/(1-\rho)^2$;
computing it is an exercise on Tutorial 2.

# Discrete Simulation of $M/M/1$

Lab 1 runs a discrete-time simulation of $M/M/1$. Because a birth-death
process is a Markov chain on a clock of length $\Delta t$, each step has
three outcomes: birth, death, or no change.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/discreteMM1State.png" style="width:70%; height:auto; display:block;">

Figure: <i>Discrete-time $M/M/1$: from occupancy $k$ the next occupancy
is $k+1$, $k-1$, or $k$.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/probOutcome.png" style="width:60%; height:auto; display:block;">

Figure: <i>Draw $U\sim\mathrm{Uniform}(0,1)$. Birth if $U<P(b)$; death
if $P(b)\le U<P(b)+P(d)$ and $k>0$; otherwise stay.</i>

The continuous $M/M/1$ is the $\Delta t\to 0$ limit of this chain, as
Coin Monopoly became a Poisson process. In the lab, take $\Delta t$ much
smaller than $1/\lambda$ and $1/\mu$, and set $P(b)=\lambda\Delta t$,
$P(d)=\mu\Delta t$.
