# Time Averages

Birth-death analysis gave the mean number in the system. Quality of
service asks for mean *delay*, and for the mean number waiting rather
than the mean number in service. Little’s formula relates those
quantities. It applies to $G/G/1$ queues, not only to $M/M/1$, once the
system is in steady state.

Write $N(t)$ for occupancy as a function of time. The time average up to
$t$ is $$
\bar{N}_t=\frac{1}{t}\int_0^t N(\tau)\,d\tau,
$$ and we assume that the steady-state mean $\langle k\rangle$ equals
$\lim_{t\to\infty}\bar{N}_t$ (Bertsekas and Gallager, 1992, pp.
~156–157).

# Little’s Formula

Under that identification, $$
\bar{N}=\lambda\bar{T}:
$$ mean number in the system equals arrival rate times mean delay.

## A Nightclub Queue

On Thursday nights the queue typically holds 20 people, and people
arrive at 40 per hour. Little’s formula gives $$
\bar{T}=\frac{\bar{N}}{\lambda}=\frac{20}{40}=\tfrac{1}{2}\text{ hour}.
$$ If twenty people are waiting and they enter at forty per hour, half
an hour is the wait one should expect. The formula does not use the
arrival or service *distributions*, only steady state. It can be applied
to the whole system or to a subsystem (the waiting line, or the server).

# Average Delay in $M/M/1$

For $M/M/1$ we already have
$\bar{N}=\rho/(1-\rho)=\lambda/(\mu-\lambda)$. Little’s formula converts
that into mean sojourn time, $$
\bar{T}=\frac{\bar{N}}{\lambda}=\frac{1}{\mu-\lambda}.
$$ This is time in the *system*, queueing plus transmission, not waiting
time alone.

## Queueing Time and Transmission Time

Packet delay splits as $T=X+W$, transmission plus waiting, so
$\bar{T}=\bar{X}+\bar{W}$. Little’s formula applies to each piece
separately.

Arrivals at the server equal arrivals at the queue if nothing is
dropped. The mean number in service is the utilisation $\rho$, and $$
\rho=\lambda\bar{X}.
$$ For exponential service $\bar{X}=1/\mu$, so $\rho=\lambda/\mu$ as
before. Utilisation is the fraction of time the server is busy, which is
also $1-P(0)$. For a packet of $L$ bits on a line of $C$ bits per
second, $X=L/C$.

For the waiting line, $$
\bar{N}_Q=\lambda\bar{W}.
$$ With $\bar{T}=\bar{X}+\bar{W}=1/\mu+\bar{W}$ we obtain $$
\bar{W}=\frac{\rho}{\mu-\lambda},\qquad \bar{N}_Q=\frac{\rho^2}{1-\rho}.
$$

# Graphical Argument

Sketch arrivals $\alpha(\tau)$ and departures $\beta(\tau)$ against
time. The area between the curves is both $\int_0^t N(\tau)\,d\tau$ and
$\sum_{i=1}^{\alpha(t)} T_i$. Divide by $t$: $$
\bar{N}_t=\bar{\lambda}_t\bar{T}_t.
$$ If a steady state exists, the $t\to\infty$ limit is
$\bar{N}=\lambda\bar{T}$.

## A Bar Queue

The bartender spends 10 percent of her time cleaning between customers,
so $\rho=0.9$. Typically three people are waiting, $\bar{N}_Q=3$. Then
$\lambda=0.9/\bar{X}$ and $3=\lambda\bar{W}$, so
$\bar{W}/\bar{X}=3/0.9=10/3$: waiting time is a little more than three
times service time.
