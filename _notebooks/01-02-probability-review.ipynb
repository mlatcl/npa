Notation: we often write $P(x,y)$ for $P(X=x,Y=y)$. Unlike a generic
bivariate function, $P(x,y)=P(y,x)$.

The product rule relates joint and conditional. The sum rule recovers a
marginal by summing out the variable you do not care about. Continuous
analogues replace sums by integrals.

Bayes is not a third axiom; it is the product rule rearranged. The
barrels example below is a small discrete inversion of the conditioning.

## Independence

<span class="editsection-bracket"

We will often assume that two or more variables are independent. That
means knowledge of one does not change the distribution of the other, $$
P(x)=P(x\mid y),
$$ so the joint factorises, $$
P(x,y)=P(x)P(y).
$$ Factorisation is what makes so much of the later analysis tractable:
superposition of Poisson processes, Kleinrock’s independence assumption,
and the product-form results for networks of queues all start here.

# Expectations

A probabilistic model summarises our beliefs about a system. To say more
we often summarise the model by its expectations.

The expected value of a function $f(x)$ under a distribution $P(x)$ is
$$
\left\langle f(x)\right\rangle_{P(x)}=\sum_{x}P(x)\,f(x).
$$ You will also see $E\{f(x)\}$ for the same quantity.

## A Small Discrete Example

Suppose $X$ takes values $0,1,2$ and $f(x)=x^2$.

| $x$        | 0   | 1   | 2   |
|:-----------|:----|:----|:----|
| $P(x)$     | 0.2 | 0.5 | 0.3 |
| $f(x)=x^2$ | 0   | 1   | 4   |

Then $$
\langle x^2\rangle = 0\times 0.2 + 1\times 0.5 + 4\times 0.3 = 1.7,
$$ and $$
\langle x\rangle = 0\times 0.2 + 1\times 0.5 + 2\times 0.3 = 1.1.
$$ The expectation of $x$ is the first moment, or *mean*: the centre of
probability mass. The expectation of $x^2$ is the second moment.

## Variance

Variance is the second moment minus the square of the mean, $$
\sigma^2 = \langle x^2\rangle - \langle x\rangle^2.
$$ It is always non-negative. Its square root is the standard deviation.
For the table above, $$
\sigma^2 = 1.7 - (1.1)^2 = 0.49, \qquad \sigma = 0.7.
$$

# Continuous Variables

So far $x$ and $y$ have been discrete. For a continuous quantity — the
height of a computer-science student — we could discretise into bins, as
the pictorial definition already did. That is awkward if we want to
answer *any* question about height. A *probability density function*
(PDF) does not give probabilities directly; we derive them from it by
integration.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/pdfOfHeight.png" style="width:60%; height:auto; display:block;">

Figure: <i>A Gaussian PDF representing heights of computer
scientists.</i>

We write a density with a small $p(\cdot)$ and a discrete distribution
with a capital $P(\cdot)$. A density is a positive function whose
integral over the region of interest is one. The Gaussian above is a
model of height. The probability that a student is taller than
$2\,\mathrm{m}$ is the integral from 2 to $\infty$. Because the total
area is one and the function is positive, that probability lies in
$(0,1)$. The probability of being *exactly* $2\,\mathrm{m}$ is zero:
that is the limit of a region whose width tends to zero.

## Manipulating Densities

The same rules apply. We take as given $$
p(y\mid x)=\frac{p(x\mid y)\,p(y)}{p(x)},
$$ with $p(x,y)=p(x\mid y)p(y)$ and $p(x)=\int p(x,y)\,dy$. Discrete and
continuous variables may be mixed. Expectations under a density use an
integral, $$
\langle f(x)\rangle_{p(x)}=\int f(x)\,p(x)\,dx.
$$

## Cumulative Distribution Functions

A common question is $P(x<y)$. The function that stores the answer for
every $y$ is the cumulative distribution function (CDF), $$
P(x<y)=\int_{-\infty}^{y} p(x)\,dx
$$ when the density is supported on the whole line, or from $0$ when the
support is the positive reals. Then $P(x>y)=1-P(x<y)$. The PDF is the
derivative of the CDF.

# Sample-Based Approximations

The expectation of $f(k)$ under $P(k)$ is
$\langle f(k)\rangle=\sum_k f(k)P(k)$. From the pictorial definition, $$
P(k)=\lim_{S\to\infty}\frac{s_k}{S},
$$ where $S$ is the number of independent trials and $s_k$ the number
that landed in state $k$. For large $S$, $$
\langle f(k)\rangle \approx \frac{1}{S}\sum_{i=1}^{S} f(k_i),
$$ with $k_i$ independent samples from $P(k)$. The *sample mean*
$\bar k = S^{-1}\sum_i k_i$ is the special case $f(k)=k$. The same
replacement of integrals by averages holds for densities.

Most texts use “mean” for both the sample mean and the mean of the
distribution. Keep the distinction. Later in the course we will also
meet time averages, and we will assume that a time average equals an
average under a steady-state distribution.

## Mean of a Poisson

The mean of $P(k\mid\mu)=\mu^k e^{-\mu}/k!$ is $\mu$. Pull
$\mu e^{-\mu}$ out of the sum, cancel $k/k!$, shift the index $m=k-1$,
and recognise the series for $e^{\mu}$. The variance is an exercise on
Tutorial Sheet 1.

```
import sys
import os
_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
_py = os.path.join(_root, 'python')
if os.path.isdir(_py) and _py not in sys.path:
    sys.path.insert(0, _py)
import numpy as np
import matplotlib.pyplot as plt
import npa_plots
```

We can check the Poisson mean by sampling. The helper
`exponential_sample` is the inverse-CDF sampler from the old MATLAB
`exponentialSamp.m`; Poisson counts are most easily drawn from NumPy.

```
mu = 2.0
rng = np.random.default_rng(1)
k = rng.poisson(mu, size=20_000)
print("sample mean", k.mean(), "sample variance", k.var())
```
