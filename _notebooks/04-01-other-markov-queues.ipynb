# Other Markov Queues

The $M/M/1$ queue is the simplest Markov queue. Two nearby members of
the same birth-death family change one of Kendall’s slots: a finite
buffer $M/M/1/m$, and an unlimited number of servers $M/M/\infty$. Both
keep Poisson arrivals and exponential service; only the rates
$\lambda_k$ and $\mu_k$ change.

# Infinite Servers: $M/M/\infty$

Sometimes it is useful to give every customer their own server.
Occupancy $k$ then has death rate $\mu_k=k\mu$ (each of the $k$ busy
servers finishes independently at rate $\mu$) and a constant birth rate
$\lambda_k=\lambda$. There is never a queue: a customer starts service
immediately.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/stateMMinf.png" style="width:80%; height:auto; display:block;">

Figure: <i>State diagram for the $M/M/\infty$ queue: unbounded $k$,
death rate rising linearly with occupancy.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/boundaryFlowMMinf.png" style="width:50%; height:auto; display:block;">

Figure: <i>Boundary flow for $M/M/\infty$:
$\lambda P(k-1)=k\mu P(k)$.</i>

Detailed balance is $$
\lambda P(k-1)=k\mu P(k),
$$ so $$
P(k)=\frac{\lambda}{k\mu}P(k-1)=P(0)\frac{a^k}{k!},\qquad a=\frac{\lambda}{\mu}.
$$ The dimensionless ratio $a$ is the *offered load*, with SI-adjacent
unit the Erlang (after A. K. Erlang). It is *not* the utilisation:
$\rho$ never exceeds 1, but $a$ for $M/M/\infty$ may be much larger than
1 because there is always a free server.

Normalisation over $k=0,1,2,\ldots$ uses
$\sum_{k=0}^{\infty}a^k/k!=e^{a}$, so $P(0)=e^{-a}$ and $$
P(k)=\frac{a^k}{k!}e^{-a}:
$$ occupancy is Poisson with mean $a$. The proportion of time the system
is in use is $1-P(0)=1-e^{-a}$.

# Finite Buffer: $M/M/1/m$

In $M/M/1$ the buffer is unbounded. A restricted buffer of $m-1$ waiting
customers, plus one in service, caps occupancy at $m$. Kendall writes
this $M/M/1/m$. Arrivals that find $k=m$ are *lost* and are assumed not
to return, so carried load is strictly less than offered load.

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/stateMM1m.png" style="width:75%; height:auto; display:block;">

Figure: <i>State diagram for $M/M/1/m$: the chain stops at $m$; there is
no transition $m\to m+1$.</i>

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/boundaryFlowMM1.png" style="width:45%; height:auto; display:block;">

Figure: <i>Interior boundary flow is the same as $M/M/1$:
$\lambda P(k-1)=\mu P(k)$.</i>

For $1\le k\le m$ the rates are those of $M/M/1$, so detailed balance is
unchanged, $$
\lambda P(k-1)=\mu P(k)\qquad\Rightarrow\qquad P(k)=\rho^k P(0),\quad\rho=\frac{\lambda}{\mu}.
$$ Normalisation now runs over only $m+1$ states, $$
\sum_{k=0}^{m}P(k)=P(0)\sum_{k=0}^{m}\rho^k=1.
$$ The inner sum is a finite geometric series, $$
\sum_{k=0}^{m}\rho^k=\frac{1-\rho^{m+1}}{1-\rho}\quad(\rho\neq 1),
$$ so $$
P(0)=\frac{1-\rho}{1-\rho^{m+1}},\qquad P(k)=\frac{\rho^k(1-\rho)}{1-\rho^{m+1}}.
$$ (If $\rho=1$ the $m+1$ states are equally likely and $P(k)=1/(m+1)$.)
Unlike $M/M/1$, a steady state exists for every $\rho$, including
$\rho\ge 1$: the finite buffer cannot grow without bound.

## Blocking Probability

An arrival is lost exactly when the system is full. PASTA (Poisson
arrivals see time averages) gives the blocking probability as the
occupancy of the last state, $$
P(m)=\frac{\rho^m(1-\rho)}{1-\rho^{m+1}}.
$$ That is the fraction of offered traffic that is rejected.
