# Erlang Loss: $M/M/m/m$

Erlang delay holds overflowing customers. Erlang *loss* does not: if all
$m$ servers are busy the arrival is rejected. Kendall writes $M/M/m/m$ —
$m$ servers and system capacity $m$, so the waiting line has length
zero. Offered load is still $a=\lambda/\mu$, but carried load $a'$ is
smaller because some traffic is blocked, as in $M/M/1/m$.

Birth-death rates are $$
\lambda_k=\begin{cases}
\lambda & 0\le k<m,\\
0 & k\ge m,
\end{cases}
\qquad
\mu_k=k\mu\quad(1\le k\le m).
$$ The value of $\mu_k$ for $k>m$ is irrelevant: those states are
unreachable.

## Truncated Poisson

Interior detailed balance is the $M/M/\infty$ relation, $$
\lambda P(k-1)=k\mu P(k)\qquad\Rightarrow\qquad P(k)=\frac{a}{k}P(k-1)=\frac{a^k}{k!}P(0).
$$ There is no $k>m$ region to match. Normalisation over $k=0,\ldots,m$
produces the *truncated Poisson* $$
P(0)=\left(\sum_{n=0}^{m}\frac{a^n}{n!}\right)^{-1},\qquad
P(k)=\frac{a^k/k!}{\sum_{n=0}^{m}a^n/n!}.
$$ As $m\to\infty$ this recovers $M/M/\infty$, as it must.

# Probability of Blocking

Waiting probability is zero: the chain never enters $k>m$. The quantity
of interest is the blocking probability, which by PASTA is $P(m)$, $$
B(m,a)=P(m)=\frac{a^m/m!}{\sum_{n=0}^{m}a^n/n!}.
$$ This is *Erlang’s first formula*, or Erlang B. The same overflow
caution as Erlang C applies; `npa_plots.erlang_b` evaluates it in log
space, matching `erlangB.m`.

# Carried Load and Utilisation

Carried load is offered load times the probability of not being blocked,
$$
a'=a\bigl(1-B(m,a)\bigr).
$$ Trunk occupancy — utilisation of a line — is then $\rho=a'/m$, not
$a/m$. In a lossless system $a'=a$ and the two coincide. The throughput
of the system, the rate at which customers actually enter service, is $$
\dot{\lambda}=\lambda\bigl(1-P(m)\bigr).
$$ Little’s formula on a lossy system must use this carried rate, not
the offered arrival rate: blocked customers never enter, so they do not
contribute to occupancy.
