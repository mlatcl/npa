# Stochastic Processes

A probabilistic process, also known as a random process or a stochastic
process, can be viewed as a \`random function’. A normal function maps a
set of inputs to a point value. A probabilistic process maps a set of
inputs to a probability distribution over values or a probability
density function. Various systems are modelled with probabilistic
processes. For example, the *Wiener process* is the process behind
Brownian motion. Stock market data is often modelled as a stochastic
process. For example the Black-Scholes formula assumes that the
logarithm of the stock market data can be modelled as a Wiener Process.
We will base our definition of a random process on that given by (Chan,
2000).

-   \*\*A\*\* random  process May be defined as a *random function*
    which, for a particular value, is a random quantity.

# Poisson Process

We’ve already met the Poisson distribution, there is a simple process
which is related to the Poisson distribution: the Poisson process. The
Poisson process gives probabilities of a number of arrivals, $k$, as a
function of time, $t$, given a \`rate’, $\lambda$.

Note the similarity between () and (). The process is simply the
distribution with $\mu$ substituted with $\lambda t$. Using this
substitution the mean of the Poisson process (the first moment of $k$)
is given by$$
\left\langle k\right\rangle =\lambda t.$$Which is the expected number of
arrivals at a given time $t$. The parameter $\lambda$ is therefore known
as the of the process. This implies that the Poisson process is a form
of counting process where the counts are to increase linearly with time
at a rate given by $\lambda$.

# Properties

In the Chapter  we will derive the Poisson process from some simple
assumptions about how arrivals occur, for the moment though we will
highlight some important properties of the Poisson process. We will
describe each in more detail below, to summarise though the fiver
properties are:

1.  Superposition of Poisson processes.
2.  Decomposition of Poisson processes.
3.  Independence of arrivals.
4.  Similarity of arrival distribution.
5.  Interarrival time distributions.

## Superposition of Poissons

The first property we will consider is superposition of Poisson flows.
It is a property of a Poisson process that if we combine $N$ independent
processes with rates $\lambda_{1},\lambda_{2},\dots,\lambda_{N}$
together the resulting process is also a Poisson process with a combined
rate, $\lambda$, equal to the sum of each of the input rates:
$\lambda=\sum_{n=1}^{N}\lambda_{n}$. This is shown pictorially in
Figure  .

## Decomposition of Poisson Flow

Just as Poisson processes can be combined to obtain a higher rate
Poisson process, they can also be decomposed to obtain lower rate
Poisson processes. Assume we have one input channel where arrivals are
occurring with a Poisson process with rate $\lambda$. This channel then
splits into $N$ separate output channels. Each input arrival is then
*independently* assigned to output channel $n$ with probability
$P\left(n\right)$. In this case the output channels then behave like
Poisson processes with rates given by
$\lambda_{n}=P\left(n\right)\lambda$. This is shown in Figure  .

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/decomposePoisson.png" style="width:80%; height:auto; display:block;">

Figure: <i>If a single Poisson process with rate $\lambda$ is switched
to different output channels, with each arrival being switched to
channel $n$ with probability $P\left(n\right)=\lambda_{n}/\lambda$, each
output channel will be a Poisson process with rate $\lambda_{n}$.</i>

## Independence of Arrivals

Consider a single Poisson process, if we look within this process at two
*disjoint* time intervals (*i.e.* periods of time that don’t overlap).
Assume we have $k_{1}$ arrivals in the first interval and $k_{2}$
arrivals in the second interval. The distribution of arrivals in the two
intervals are independent of each other,
\*i.e.\*$P\left(k_{1},k_{2}\right)=P\left(k_{1}\right)P\left(k_{2}\right)$.

## Similarity of Arrival Distributions for Identical Intervals

Consider a single Poisson process with rate $\lambda$, let the number of
arrivals between time $s$ and $s+t$ be given by $k_{1}$. Now consider a
second independent Poisson process with the same rate $\lambda$, let the
number of arrivals between time $r$ and $r+t$ be given by $k_{2}$. The
distribution of $k_{1}$ and $k_{2}$ are then equal, $$
P\left(k_{1}\right)=P\left(k_{2}\right).$$

## Interarrival Times and Poisson Processes

The Poisson process is a simple model of arrivals that is widely used
due to its analytical properties. We have already seen how a
superpositions and decompositions of Poisson processes also lead to
Poisson processes. Another attractive analytical property is the ease
with which we can compute the *interarrival time distribution* for the
Poisson process. The interarrival time distribution is the distribution
of times between arrivals. Consider waiting for the next bus at a bus
stop. If you *just missed a bus* how long will you have to wait for the
next bus to arrive? If the buses arrive as a Poisson process, then it is
straightforward to work out the distribution of your waiting time. The
interarrival time distribution for the Poisson process turns out to be
the *exponential distribution*,$$
p\left(X\right)=\lambda e^{-\lambda X}.$$ This is derived in the
examples at the end of this chapter. Stating that interarrival times are
distributed exponentially is equivalent to stating that arrivals occur
as a Poisson process. We will therefore use these formulations
interchangeably, *i.e.*when we state that the interarrival times for
students registering at the Goodwin Sports centre is given by an
exponential distribution with parameter $\lambda=2$ we are equivalently
stating that these students are arriving according to a Poisson process
with rate parameter $\lambda=2$.

# The Exponential Distribution

In the last section we saw that the exponential distribution gives can
be derived by considering the interarrival times of the Poisson
distribution. Here we give a few properties of the exponential that will
be relevant in this course.

## The Memoryless property

We have already seen in our derivation of the exponential as
interarrival time distribution that it is *memoryless*. Formally
speaking this means that for exponentially distributed variable $X$$$
P\left(X>s+t|X>t\right)=P\left(X>s\right)$$or in English, the
probability of the waiting time being greater than $s+t$ given that the
waiting time is already greater than $t$ is the same as the probability
of the waiting time being greater than $s$.

## Mean and Variance

Recall the exponential distribution $$
p(X)=\lambda e^{-\lambda X}$$ The mean of an exponential can be found to
be, $$
\left\langle X\right\rangle _{p(X)}=\frac{1}{\lambda}$$ and the variance
is $$
\left\langle X^{2}\right\rangle _{p(X)}-\left\langle X\right\rangle _{p(X)}^{2}=\frac{1}{\lambda^{2}}.$$
These are easily confirmed — they are part of Tutorial 1.

## Answer

Firstly we define the number of arrivals at time $t$ to be given by the
function $N\left(t\right)$. Next we note that the cumulative density
function (CDF) of the interarrival time distribution will be given by
$P\left(X_{i}\leq t\right)$ (see Section  for a reminder on CDFs). We
can re-write this probability as$$
P\left(X_{i}\leq t\right)=1-P\left(X_{i}>t\right),$$so, rather than
focusing on $P\left(X_{i}\leq t\right)$, we can consider the
distribution $P\left(X_{i}>t\right)$. Now we rewrite the condition in
this distribution as
(X\_{i}\>t\|T\_{i-1}=t\_{i-1})(N(t\_{i-1}+t)-N(t\_{i-1})=0).$$The right hand side of this statement expresses the condition that no arrivals occur between time $t_{i-1}=T_{i-1}$ and time $t$.
  \footnote{The `$\equiv$' sign should be read as `is equivalent to'.}
  $$
P(X\_{i}\>t\|T\_{i-1}=t\_{i-1})=P(N(t\_{i-1}+t)-N(t\_{i-1})=0)$$now, because of the similarity of the arrival distributions for intervals of identical lengths (Section\  \ref{sub:poissonSimilarities}) we can write$$
P(N(t\_{i-1}+t)-N(t\_{i-1}))=P(N(t))$$which means that$$
P(N(t\_{i-1}+t)-N(t\_{i-1})=0)=P(N(t)=0)$$and substituting $N\left(t\right)=k=0$ into (\ref{eq:process}) we find that the probability of their being no arrivals at time $t$, $P\left(N\left(t\right)=0\right)$, is $$
