# How this lab sits in the course

This is the Monday of week 3, immediately after *Birth-Death Processes
and the $M/M/1$ Queue* (lectures 4 and 5) and *Little’s Formula*
(lecture 6). The lectures derived $P(k)=(1-\rho)\rho^k$ from detailed
balance and showed a discrete-time chain with $\Delta t$ as a
computational stand-in for the continuous process. This lab asks you to
*write* that chain, sample it, and see when the sample mean is a
trustworthy estimate of $\bar{N}=\rho/(1-\rho)$. Tutorial 2, later the
same week, returns to occupancy variance and Little on paper.

# 1. Run the chain

Code a simulation with birth rate $\lambda=1$ for all $k=0,1,\ldots$ and
death rate $\mu=2$ for $k=1,1,\ldots$ (with $\mu_0=0$). At each step the
probability of $+1$ is $\lambda\Delta t$, of $-1$ is $\mu\Delta t$, and
of no change is $1-(\lambda+\mu)\Delta t$. Run for $1000$ seconds
($100{,}000$ iterations). Store the occupancy at every step: those
stored values are samples from the model the queue represents.

*While you wait for simulations to run you might like to start looking
at the questions below.*

```
occupancy = npa_plots.simulate_mm1(1.0, 2.0, delta_t=0.01, n_iters=100_000, rng=1)
print(occupancy[:20], occupancy.mean())
```
