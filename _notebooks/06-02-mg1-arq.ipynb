# Automatic Repeat Request (ARQ) System.

The $M/G/1$ queue system can be used for modelling various systems of
interest. In an ARQ system packets are transmitted and the system waits
for an acknowledgement. If no acknowledgement is received the system
retransmits. In this section we review Example  3.15 from (Bertsekas and
Gallager, 1992). A schematic of the system is given in Figure  . For the
purposes of this analysis, we will assume

<img class="" src="https://mlatcl.github.io/npa/slides/diagrams/npa/ARQ.png" style="width:80%; height:auto; display:block;">

Figure: <i>The ARQ System.</i>

1.  That each packet takes 1 millisecond to be transmitted.
2.  The system waits a maximum of $n-1$ milliseconds for an
    acknowledgement. If there is no acknowledgement the system
    retransmits.
3.  The acknowledgement occurs if the packet is received without errors.
    Otherwise acknowledgement doesn’t occur.
4.  The probability of an error in a packet is taken to be $p$ and we
    take errors to occur independently.
5.  That packets arrive as a Poisson process with rate $\lambda$.

In practice, the fourth assumption (independence of errors) is probably
unrealistic . If an error occurs due to problems in the transmission
channel, it is likely that several packets in a row will not be
received, so the errors do not occur independently. The time interval
between transmission and re-transmission will be $1+kn$ milliseconds
with probability $\left(1-p\right)p^{k}$. This time interval is
associated with $k$ retransmissions and one successful transmission. The
probability of this being the service time is given by$$
P\left(X=1+kn\right)=\left(1-p\right)p^{k}$$which is recognised as a
*geometric distribution*.

## Average Service Time

Using this distribution we can compute the expected service time which
is given by since $\sum_{k=0}^{\infty}p^{k}=\frac{1}{1-p}$ and
$\sum_{k=0}^{\infty}kp^{k}=\frac{p}{\left(1-p\right)^{2}}$ .

## ARQ Utilization Factor

Strictly speaking, it probably doesn’t make sense to use the term
`utilization factor' for the ARQ system. This is because during a large portion of the wait it is not actually`utilizing’
the channel, it is merely waiting for an acknowledgement. However we can
still consider,
\$=X$, to be the proportion of time that the system will be `busy'. This `busy factor',$\$
=+n,\$\$obviously still has to be less than 1, otherwise the steady
state would not exist. This busy time is affected by three parameters:
the rate of arrivals, $\lambda$, the amount of time that the system
waits for an acknowledgement, $n$, and the probability of an error, $p$.
Note that as $p\rightarrow1$ the second term $\rightarrow\infty$.

## ARQ Second Moment

The second moment of the service time issince $$
\sum_{k=0}^{\infty}k^{2}p^{k}=\frac{p+p^{2}}{\left(1-p\right)^{3}}$$
These moments can be used with the Pollaczek-Khinchin formula to obtain
statistics for the queue. Examples

    of staff serving. When there are no customers at the bar, the staff
    member leaves the bar to check on the kitchen. Assuming these checks
    are exponentially distributed in length, with mean $\bar{V}$ and
    that customer service times are exponentially distributed with mean
    $\bar{X}$ what is the average waiting time in the queue? \[0\]
    Exponential standard deviation = $\frac{1}{\mu}$,
    mean$=\frac{1}{\mu}$. So if mean is $\bar{V}$, standard deviation is
    $\bar{V}$ and variance is $\bar{V}^{2}$. Thus second moment is
    $\left\langle V^{2}\right\rangle =2\bar{V}^{2}$. Similarly for
    second moment of service time
    $\left\langle X^{2}\right\rangle =2\bar{X}^{2}$. Defining
    $\rho=\lambda\bar{X}$ we have $$
